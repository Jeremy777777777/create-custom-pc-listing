# Listing Testing environment smoke test
Select **Python (Listing Testing)**. This notebook tests in-memory objects only; it does not modify product files.

```python
import sys
from pathlib import Path
from importlib.metadata import version
assert Path(sys.prefix).name == 'listing-testing', f'Wrong environment: {sys.prefix}'
assert sys.version_info[:2] == (3, 12), sys.version
print(sys.executable)
for package in ('Pillow', 'openpyxl', 'PyYAML', 'pytest', 'jupyterlab', 'ipykernel', 'nbclient'):
    print(package, version(package))
```

```python
from io import BytesIO
from PIL import Image
from openpyxl import Workbook, load_workbook
import yaml
assert yaml.safe_load('style: B17')['style'] == 'B17'
image_buffer = BytesIO()
Image.new('RGB', (2000, 2000), 'white').save(image_buffer, format='PNG')
image_buffer.seek(0)
assert Image.open(image_buffer).size == (2000, 2000)
book = Workbook()
book.active['A1'] = 'Listing smoke test'
book.active['B1'] = 16
book_buffer = BytesIO()
book.save(book_buffer)
book_buffer.seek(0)
restored = load_workbook(book_buffer)
assert restored.active['A1'].value == 'Listing smoke test'
assert restored.active['B1'].value == 16
print('PASS: YAML, 2000px PNG and Excel round trips')
```
